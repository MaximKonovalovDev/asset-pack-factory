# 3D Router - Colab Worker Node
Connects this Google Colab GPU runtime to the local 3D Router gateway.
- Runtime: Free T4 GPU (12h/session)
- Purpose: 3D generation, Blender rendering, mesh baking & upscale
- Callback: Phones home via persistent gateway tunnel


```python
GDR_WORKER_ID = "colab-1"
GDR_WORKER_TOKEN = "colab-secret-token"
GDR_TUNNEL_URL = "https://pharmacy-sharp-uniprotkb-pulling.trycloudflare.com"

import urllib.request, urllib.error, json, time, os, sys

def register(disconnect=False):
    payload = {'accountId': GDR_WORKER_ID, 'token': GDR_WORKER_TOKEN, 'capabilities': {'gpu': True, 't4': True}}
    if disconnect: payload['disconnect'] = True
    req = urllib.request.Request(f"{GDR_TUNNEL_URL}/api/tunnel", data=json.dumps(payload).encode(), headers={'Content-Type': 'application/json'})
    try:
        with urllib.request.urlopen(req, timeout=10) as r:
            print('Worker registered with 3D Router gateway.')
    except Exception as e:
        print(f'Registration warning: {e}')

register()
print(f'Colab Worker active. Polling {GDR_TUNNEL_URL} for 3D tasks...')
```
